# Simple Data Filtering

## Objective

To create a collection of student records and filter students based on conditions such as marks, age, and department.

```python
students = [
    {
        "id": 101,
        "name": "Arun",
        "age": 20,
        "department": "MCA",
        "marks": 85
    },
    {
        "id": 102,
        "name": "Dharani",
        "age": 21,
        "department": "MCA",
        "marks": 90
    },
    {
        "id": 103,
        "name": "Kavin",
        "age": 20,
        "department": "MCA",
        "marks": 78
    },
    {
        "id": 104,
        "name": "Premi",
        "age": 21,
        "department": "MCA",
        "marks": 88
    },
    {
        "id": 105,
        "name": "Sowmini",
        "age": 22,
        "department": "BCA",
        "marks": 72
    },
    {
        "id": 106,
        "name": "Rohitha",
        "age": 21,
        "department": "MCA",
        "marks": 91
    },
    {
        "id": 107,
        "name": "Vijay",
        "age": 22,
        "department": "BCA",
        "marks": 68
    },
    {
        "id": 108,
        "name": "Keerthi",
        "age": 20,
        "department": "MCA",
        "marks": 95
    }
]

print(students)
```

```python
high_mark_students = []

for student in students:
    if student["marks"] >= 80:
        high_mark_students.append(student)

print("Students with marks 80 or above:")
for student in high_mark_students:
    print(student)
```

```python
mca_students = []

for student in students:
    if student["department"] == "MCA":
        mca_students.append(student)

print("Students from MCA Department:")
for student in mca_students:
    print(student)
```

```python
age_filtered_students = []

for student in students:
    if student["age"] >= 21:
        age_filtered_students.append(student)

print("Students aged 21 or above:")
for student in age_filtered_students:
    print(student)
```

```python
mca_high_mark_students = []

for student in students:
    if student["department"] == "MCA" and student["marks"] >= 85:
        mca_high_mark_students.append(student)

print("MCA Students with marks 85 or above:")

for student in mca_high_mark_students:
    print(student)
```

## Explanation of Results

The student records were stored using a list of dictionaries.

Different filtering conditions were applied to analyze the student data.

Students with marks 80 or above were identified successfully.

Students belonging to the MCA department were filtered successfully.

Students aged 21 or above were also identified.

A multiple-condition filter was used to find MCA students who scored 85 or above.

This task helped demonstrate how Python can be used to filter and analyze structured data using loops, conditions, and logical operators.
